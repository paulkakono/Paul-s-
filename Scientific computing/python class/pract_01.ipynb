```python
import pandas as pd

student_data = {
    "Name": ["jane", "Joseph", "Simon"],
    "Maths": [23, 49, 59],
    "Age": [23, 49, 59],
    "English": [23, 49, 59],
}

students = pd.DataFrame(student_data)

maths_range = students["Maths"].max() - students["Maths"].min()
print(maths_range)
```
