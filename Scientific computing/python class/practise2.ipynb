```python
s1=int(input("Enter a value for s1:"))
s2=int(input("Enter a value for s2:"))
s3=int(input("Enter a value for s3:"))
s4=int(input("Enter a value for s4:"))
s5=int(input("Enter a value for s5:"))
avg=(s1+s2+s3+s4+s5)/5
print('Average of five subject marks is: '+str(avg))
if(avg>=80):
    print('You are an outstanding student.')
if(avg>=70 and avg<80) :
    print('You are a good student.')
if(avg>=60 and avg<70) :
    print('You are an average student.')
if(avg>=50 and avg<60) :
    print('You are a below-average student.')
   
    
       
```
